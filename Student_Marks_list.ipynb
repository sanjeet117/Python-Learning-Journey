```python
marks = [78, 85, 90, 67, 88, 92, 76]
```

```python
# Print the complete list
print(marks)
```

```python
# Print the first element
print(marks[0])
```

```python
# Print the last element
print(marks[-1])
```

```python
# Print elements from index 2 to 5
print(marks[2:6])
```

```python
# Find the number of elements
print(len(marks))
```

```python
# Find maximum marks
print(max(marks))
```

```python
# Find minimum marks
print(min(marks))
```

```python
# Find total marks
print(sum(marks))
```

```python
# Sort marks in ascending order
marks.sort()
print(marks)
```

```python
# Sort marks in descending order
marks.sort(reverse = True)
print(marks)
```

```python
# Add 95
marks.append(95)
print(marks)
```

```python
# Add [81, 84]
marks.extend([81,84])
print(marks)
```

```python
# Remove 67
marks.remove(67)
print(marks)
```

```python
# Count how many times 90 occurs
print(marks.count(90))
```

```python
# Find the index of 88
print(marks.index(88))
```
