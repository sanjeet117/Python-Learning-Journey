```python
# 1. String to Integer
res_int = int("100")
print(f'int("100") -> {res_int} ({type(res_int)})')
print("Explanation: The string '100' is parsed into the integer value 100.\n")
```

```python
# 2. String to Float
res_float = float("45.67")
print(f'float("45.67") -> {res_float} ({type(res_float)})')
print("Explanation: The string '45.67' is converted to a decimal floating-point number.\n")
```

```python
# 3. Integer to String
res_str = str(500)
print(f'str(500) -> {repr(res_str)} ({type(res_str)})')
print("Explanation: The integer 500 is converted into text representation '500'.\n")
```

```python
# 4. Integer to Boolean
res_bool = bool(1)
print(f'bool(1) -> {res_bool} ({type(res_bool)})')
print("Explanation: Any non-zero integer evaluates to True in boolean context.\n")
```

```python
# 5. Tuple to List
res_list = list((1, 2, 3))
print(f'list((1, 2, 3)) -> {res_list} ({type(res_list)})')
print("Explanation: The immutable tuple is converted into a mutable list containing the same elements.\n")
```

```python
# 6. List to Tuple
res_tuple = tuple([1, 2, 3])
print(f'tuple([1, 2, 3]) -> {res_tuple} ({type(res_tuple)})')
print("Explanation: The mutable list is converted into an immutable tuple.\n")
```

```python
# 7. List to Set (Deduplication)
res_set = set([1, 2, 2, 3])
print(f'set([1, 2, 2, 3]) -> {res_set} ({type(res_set)})')
print("Explanation: The list is converted to an unordered set, automatically removing the duplicate value 2.\n")
```
