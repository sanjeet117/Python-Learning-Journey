```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}

java_students = {"Priya", "Karan", "Rohit", "Simran"}
```

```python
# Print both sets
print(python_students)
print(java_students)
```

```python
# Find students learning either Python or Java
either_students = python_students.union(java_students)
print('students learning either python or java:',either_students)
```

```python
# Find students learning both Python and Java
both_students = python_students.intersection(java_students)
print(both_students)
```

```python
# Find students learning only Python
only_python_students = python_students.difference(java_students)
print(only_python_students)
```

```python
# Find students learning only Java
only_java_students = java_students.difference(python_students)
print(only_java_students)
```

```python
# Find students belonging to exactly one group
exactly_one_group = python_students.symmetric_difference(java_students)
print(exactly_one_group)
```

```python
# Add a new student
python_students.add('Riya')
java_students.add('Sonal')
print(python_students)
print(java_students)
```

```python
# Remove a student
python_students.remove('Riya')
java_students.remove('Sonal')
print(python_students)
print(java_students)
```

```python
# Demonstrate discard()
# discard() removes an item from a set if it exists, and does nothing (no error) if it doesn’t
a = {1,2,3,4,5,1,2}
a.discard(1)
a.discard(7)
print(a)
```

```python
a.remove(4)
print(a)
a.remove(9)
```
