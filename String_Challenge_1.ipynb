```python
message = " Welcome To Python Programming Class "
```

```python
# Remove extra spaces
message1 = message.strip()
print(message)
print(message1)
```

```python
# Convert everything to lowercase
message2 = message.lower()
print(message)
print(message2)
```

```python
# Convert everything to uppercase
message3 = message.upper()
print(message)
print(message3)
```

```python
# Convert to title case
message4 = message.title()
print(message)
print(message4)
```

```python
# Replace "Python" with "Advanced Python"
message5 = message.replace('Python','Advanced Python')
print(message)
print(message5)
```

```python
# Check whether the string starts with "Welcome"
message6 = message.find('Welcome')
message7 = message.startswith('Welcome')
print(message)
print(message6)
print(message7)
```

```python
# Check whether it ends with "Class"
message8 = message.endswith('Class')
message9 = message.endswith('Class ')
print(message)
print(message8)
print(message9)
```

```python
# Count occurrences of "o"
print(message.count('o'))
print(message)
```

```python
# Find the position of "Programming"
print(message.find('Programming'))
print(message)
```

```python
# Split the sentence into words
print(message.split())
print(message)
```
