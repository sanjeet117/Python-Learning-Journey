```python
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

```python
# Display all products
print(cart)
```

```python
# Access first and last products
print('first product:',cart[0])
print('last product:',cart[-1])
```

```python
# Add "Webcam"
cart.append('Webcam')
print(cart)
```

```python
# Insert "USB Hub" at index 2
cart.insert(2,'USB Hub')
print(cart)
```

```python
# Remove "Mouse"
cart.remove('Mouse')
print(cart)
```

```python
# Remove the last item using pop()
cart.pop()
print(cart)
```

```python
# Find the index of "Monitor"
print(cart.index('Monitor'))
```

```python
# Count occurrences of "Laptop"
print(cart.count('Laptop'))
```

```python
# Create a copy of the cart
cart_copy = cart.copy()
print(cart_copy)
```

```python
# Reverse the cart
cart.reverse()
print(cart)
```

```python
# Sort the products alphabetically
cart.sort()
print(cart)
```

```python
'''Explain the difference between

append() 

extend()

insert()

remove()

pop()

clear()

copy()

sort()

reverse()'''

'''append(x)  Put one thing at the end of the list.
Example: cart.append('Mouse') → adds 'Mouse' as the last item.

extend([x, y, ...]) Add many things at the end, one by one.
Example: cart.extend(['Mouse', 'Cable']) → adds both items to the end.

insert(i, x) Put one thing at a specific position.
Example: cart.insert(1, 'Mouse') → puts 'Mouse' at index 1 (second place)


remove(x)  Remove the first box that has this value.
Example: cart.remove('Mouse') → deletes the first 'Mouse' it finds.
If it's not there, you get an error.

pop(i) Remove the item at position i.

cart.pop() → removes and returns the last item.

cart.pop(2) → removes and returns the item at index 2

clear() Remove everything from the list. List becomes empty [].

copy() Make a new list with the same items.
Example: new_cart = cart.copy() → new_cart is a separate list with same contents

sort() Rearrange the list into order (numbers: small to big; words: A to Z).
Example: nums.sort() → [3, 1, 2] becomes [1, 2, 3].
This changes the original list.

reverse()  Flip the list backwards.
Example: cart.reverse() → ['A', 'B', 'C'] becomes ['C', 'B', 'A'].
This also changes the original list and returns None '''
```
