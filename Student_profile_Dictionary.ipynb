```python
student = {

"name": "Rahul",

"age": 22,

"course": "Python",

"city": "Bangalore",

"marks": 88

}
```

```python
# Print the complete dictionary
print(student)
print(type(student))
```

```python
# Print the student's name
print(student['name'])
```

```python
# Print their course
print(student['course'])
```

```python
# Print all keys
print(student.keys())
```

```python
# Print all values
print(student.values())
```

```python
# Print all key-value pairs
print(student.items())
```

```python
# Change marks from 88 to 92
student.update({'marks':92})
print(student.values())
```

```python
# Add "email"
student.update({'email':'sanjeet@gmail.com'})
print(student.items())
```

```python
# Add "phone"
student['phone'] = 12345
print(student.items())
```

```python
# Remove "city"
student.pop('city')
print(student.keys())
```

```python
# Use get() to retrieve "name"
print(student.get('name'))
```

```python
# Create a copy of the dictionary
student_copy = student.copy()
print(student_copy)
```

```python
'''
Explain

keys() :- returns all the keys (labels) in the dictionary. print(student.keys())

values():- returns all the values in the dictionary. print(student.values())

items():- returns all key value pairs as tuples. print(student.items()).Use: When you want to loop over both keys and values together.

get():- it is used to get the value of a key safely. If the key is not present, it returns None (or a default value you give).
        print(student.get("name"))     Output: Aarav
        print(student.get("age"))      Output: None
        print(student.get("age", 0))   Output: 0  (default value)

update():- it is used to add or change many keys at once.

pop():- removes a key and returns its value.

copy():- creates a new dictionary with the same key value pairs.  
'''
```
