```python
laptop = {

"brand": "Dell",

"model": "XPS 15",

"price": 120000,

"ram": "16GB",

"storage": "512GB SSD",

"available": True

}
```

```python
# Print brand
print(laptop['brand'])
```

```python
# Print model
print(laptop['model'])
```

```python
# Print price
print(laptop['price'])
```

```python
# Change price
laptop['price'] = 110000
print(laptop['price'])
```

```python
# Add processor information
laptop['processor'] = 'Ryzon5'
laptop.items()
```

```python
# Add GPU information
laptop['GPU'] = 'RTX4090'
laptop.items()
```

```python
# Change RAM to "32GB"
laptop['ram'] = '32GB'
```

```python
# Remove "available"
laptop.pop('available')
laptop.items()
```

```python
# Print all keys
print(laptop.keys())
```

```python
# Print all values
print(laptop.values())
```

```python
# Print all items
print(laptop.items())
```
