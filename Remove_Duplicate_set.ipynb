```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
```

```python
# Print the original list
print(numbers)
```

```python
# Convert the list into a set
set_numbers = set(numbers)
print(set_numbers)
```

```python
# Observe which duplicates disappear
print(numbers)
print(set_numbers)
```

```python
# Convert the set back into a list
numbers2 = list(set_numbers)
print(numbers2)
```

```python
# Print the number of original elements
print('Number of original elements:', len(numbers))
```

```python
# Print the number of unique elements
print('Number of unique elements:', len(set_numbers))
```

```python
# Create another example using duplicate student names.
students = ['Rohit','Ajay','Sunil','Rohit','Suny','Deepak','Ajay']
print(students)
set_students = set(students)
print(set_students)
```

```python
# Explain why sets are useful when working with duplicate data.
'''Set are useful for removing duplicates data from a list.sets are built for quick lookups.processing with duplicate data take time and 
take more memory. Using sets can help to reduce the time and memory required to process data with duplicates.'''
```
