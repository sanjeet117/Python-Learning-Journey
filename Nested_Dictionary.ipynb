```python
employee = {

"name": "Amit",

"department": "Engineering",

"skills": {

"language": "Python",

"database": "PostgreSQL",

"cloud": "AWS"

},

"salary": 80000

}
```

```python
# Print employee name
print(employee['name'])
```

```python
# Print department
print(employee['department'])
```

```python
# Print complete skills dictionary
print(employee['skills'])
```

```python
# Print programming language
print(employee['skills']['language'])
```

```python
# Print database
print(employee['skills']['database'])
```

```python
# Print cloud technology
print(employee['skills']['cloud'])
```

```python
# Change "Python" to "Python + JavaScript"
employee['skills']['language'] = 'Python + JavaScript'
print(employee['skills']['language'])
print(employee.items())
```

```python
# Change salary
employee['salary'] = 100000
print(employee['salary'])
```

```python
# Add "experience": 3
employee['experience'] = 3
print(employee.items())
```

```python
# Add another skill under the skills dictionary
employee['skills']['OS'] = 'linux'
print(employee['skills']['OS'])
```

```python
# Explain how nested dictionaries are accessed.
'''
Nested dictionaries are accessed by chaining keys: you use one key to get the inner dictionary,
then another key to get the value inside it '''
print(employee['skills']['language'])   # example of nested dictionaries
```
