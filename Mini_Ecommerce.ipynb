```python
products = [

{

"name": "Laptop",

"price": 70000,

"brand": "Dell"

},

{

"name": "Phone",

"price": 40000,

"brand": "Samsung"

},

{

"name": "Tablet",

"price": 30000,

"brand": "Apple"

}

]
```

```python
# Print all products
print(products)
```

```python
# Print first product
print(products[0])
```

```python
# Print second product's price
print(products[1]['price'])
```

```python
# Print third product's brand
print(products[2]['brand'])
```

```python
# Change first product's price
products[0]['price'] = 80000
print(products[0])
```

```python
# Add "rating" to the second product
products[1]['rating'] = 7
print(products[1])
```

```python
# Add another product manually
products.append({'name':'Desktop','price':35000,'brand':'Dell'})
print(products[-1])
```

```python
# Print the final dataset
print(products)
```
